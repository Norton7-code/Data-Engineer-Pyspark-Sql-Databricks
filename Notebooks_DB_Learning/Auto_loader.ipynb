### Autoloader(Streaming Dataframe)

```python
%python
# setting up autoloder python

df = spark.readStream.format("cloudFiles") \
  .option("cloudFiles.format", "parquet") \
  .option("cloudFiles.schemaLocation","s3://inboundfiless/Outbound/Checkpoint") \
  .option("cloudFiles.allowOverwrites", "false") \
  .option("cloudFiles.schemaEvolutionMode", "rescue") \
  .load("s3://inboundfiless/Inbound/")
```

```python
%sql
use catalog ingestion_catalog;
use schema inbound_ingestion;
```

```python
df.writeStream.format("delta").option("checkpointLocation", "s3://inboundfiless/Outbound/Checkpoint").outputMode("append").trigger(availableNow = True).start("s3://inboundfiless/Outbound/")
```

### Autoloader(streaming table)

```python
df = spark.readStream.format("cloudFiles") \
  .option("cloudFiles.format", "parquet") \
  .option("cloudFiles.schemaLocation","s3://inboundfiless/Outbound/Checkpoint") \
  .option("cloudFiles.schemaEvolutionMode", "AddNewColumns") \
  .load("s3://inboundfiless/Inbound/")
```

```python
df.writeStream.format("delta").option("checkpointLocation", "s3://inboundfiless/Outbound/Checkpoint").outputMode("append").trigger(availableNow = True).toTable("ingestion_catalog.inbound_ingestion.target_table")
```
