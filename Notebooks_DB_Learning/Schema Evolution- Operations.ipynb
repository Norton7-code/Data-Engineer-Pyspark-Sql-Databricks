## reading from the external location file and writing to a table

```python
df = spark.read.format("csv").option("header", "true").option("InferSchema", "true").load("s3://inboundfiless/Source_Files/credit_card_transcations.csv")
```

```python
df.union(spark.createDataFrame([11111,'hyderabad','20-oct-25','axis','loan','M',99282]))
```

```python
from pyspark.sql.functions import *
df = df.withColumn("flag",lit("manual"))
```

```python
df.display()
```

##before merge schema true

```python
df.write.mode("overwrite").format("delta").saveAsTable("ingestion_catalog.inbound_ingestion_local.credit_card_details")
```

##After Merge schema True

```python
%python

df.write.mode("overwrite").format("delta").option("mergeSchema",True).saveAsTable("ingestion_catalog.inbound_ingestion_local.credit_card_details")
```
