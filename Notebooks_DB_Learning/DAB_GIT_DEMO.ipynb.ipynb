```python

df = spark.read.format("csv").options(header='true', inferSchema='true').load("/Volumes/ingestion_catalog/streaming_dlt/streaming_sources/train.csv")
display(df)
```

```python
df.write.mode("append").format("Delta").saveAsTable("dab_demo.cert.Trains")
```

```python
from pyspark.sql.functions import *
df = df.groupBy("train_name").agg(sum("seats_available").alias("total_Seats"))
df.display()
```

```python
df.write.mode("append").format("Delta").saveAsTable("dab_demo.cert.Trains_Seats")
```
