```python
df = spark.read.table("dab_demo.dev.hospital_data")
```

```python

from pyspark.sql.functions import *
from pyspark.sql.types import *
```

```python
df = df.withColumn("NumberOfdays_stayed",datediff(col("discharge_date"),col("admission_date")))
df.display()
```

```python
df = df.groupBy("diagnosis").agg(count("patient_id").alias("count"))
df.display()
```

```python
df.write.mode("append").saveAsTable("dab_demo.dev.diagnosis_stats")
```
