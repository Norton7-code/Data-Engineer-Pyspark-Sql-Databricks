```python
df = spark.read.format("csv").option("header", "true").load("s3://amzndeltalake/hospital.csv")
display(df)
```

```python
df.write.format("delta").option("mergeSchema", "true").save("s3://amzndeltalake/delta")
```

```python
%python
# setting up autoloder python

df_streaming = spark.readStream.format("cloudFiles") \
  .option("cloudFiles.format", "csv") \
  .option("cloudFiles.schemaLocation","s3://amzndeltalake/schema") \
  .option("cloudFiles.checkpointLocation", "s3://amzndeltalake/schema") \
  .option("cloudFiles.allowOverwrites", "false") \
  .option("cloudFiles.schemaEvolutionMode", "rescue") \
  .load("s3://amzndeltalake/")


```

```python
df_streaming.writeStream.outputMode("append").format("delta").option("checkpointLocation", "s3://amzndeltalake/schema").trigger(availableNow=True).start("s3://amzndeltalake/STAGE")
```
